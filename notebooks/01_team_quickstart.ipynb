### 1. Visualize the 4 Network Topologies ($N=10$)

```python
topologies = ['ring', 'grid', 'erdos_renyi', 'scale_free']
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, topo in zip(axes, topologies):
    G = generate_topology(topo, num_nodes=10, seed=42)
    stats = compute_graph_metrics(G)
    nx.draw(G, ax=ax, with_labels=True, node_color='lightblue', edge_color='gray')
    ax.set_title(f"{topo.capitalize()}\nDiam: {stats['diameter']}, Fiedler: {stats['algebraic_connectivity']:.2f}")

plt.tight_layout()
plt.show()
```

### 2. Step Through the Consensus Environment

```python
env = NetworkedConsensusEnv(num_agents=6, topology_type='ring', max_steps=10)
obs, info = env.reset(seed=42)
print(f"Initial votes: {env.votes}")
print(f"Consensus reached: {info['consensus']}")

# Step 1: Flip vote of agent_0
actions = {agent: 0 for agent in env.agents}
actions['agent_0'] = 1
next_obs, rewards, terms, truncs, step_info = env.step(actions)

print(f"Post-action votes: {env.votes}")
print(f"Step reward: {rewards['agent_0']:.2f}")
```
